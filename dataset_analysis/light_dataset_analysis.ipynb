```python
import pandas as pd
from itertools import product
```

```python
data = pd.read_csv("../connections_princeton.csv")

neuropil_types = list(data["neuropil"].unique())

nt_types = list(data["nt_type"].unique())
```

```python
print(f"all neuropil types: \n {neuropil_types}")
print(f"all nt types: \n {nt_types}")
```

```python
print("possible pair subgraphs: ")

pil_nt_combinations = list(product(neuropil_types, nt_types))

print(pil_nt_combinations)
print(f"number of possible combinations: {len(pil_nt_combinations)}")
```

Analyze the relationship between these two (i.e. which combination frequently appears together). We can then look at the cycles related to that combination.

```python
connection_dict = {f"{pil}-{nt}": None for (pil,nt) in pil_nt_combinations}

for (pil, nt) in pil_nt_combinations:
    connections = data[(data["neuropil"] == pil) & (data["nt_type"] == nt)]
    print(f"At {pil}-{nt} subgraph there are {len(connections)} connections within this graph")
```
